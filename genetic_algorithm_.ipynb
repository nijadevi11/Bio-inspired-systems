```python
import random

# Number of community members
members = 5

# Resource demand for each member
water_demand = [100, 120, 80, 150, 110]
electricity_demand = [50, 60, 40, 70, 55]
food_demand = [30, 40, 25, 50, 35]

# Cost per unit of resource
water_cost = 2
electricity_cost = 5
food_cost = 3


# Create one possible solution
def create_solution():
    solution = []

    for i in range(members):
        water = random.uniform(0, water_demand[i])
        electricity = random.uniform(0, electricity_demand[i])
        food = random.uniform(0, food_demand[i])

        solution.append([water, electricity, food])

    return solution


# Calculate fitness
def fitness(solution):

    total_demand = 0
    total_allocated = 0
    total_cost = 0

    for i in range(members):

        total_demand += (
            water_demand[i]
            + electricity_demand[i]
            + food_demand[i]
        )

        total_allocated += sum(solution[i])

        total_cost += (
            solution[i][0] * water_cost
            + solution[i][1] * electricity_cost
            + solution[i][2] * food_cost
        )

    # Resource utilization
    utilization = total_allocated / total_demand

    # Lower cost is better
    cost_score = 1 / (1 + total_cost)

    # Combine objectives
    fitness_value = (
        0.6 * utilization +
        0.4 * cost_score
    )

    return fitness_value


# Selection
def selection(population):

    population.sort(
        key=fitness,
        reverse=True
    )

    return population[:10]


# Crossover
def crossover(parent1, parent2):

    child = []

    for i in range(members):

        if random.random() < 0.5:
            child.append(parent1[i].copy())
        else:
            child.append(parent2[i].copy())

    return child


# Mutation
def mutation(solution):

    for i in range(members):

        if random.random() < 0.1:

            solution[i][0] = random.uniform(
                0, water_demand[i]
            )

            solution[i][1] = random.uniform(
                0, electricity_demand[i]
            )

            solution[i][2] = random.uniform(
                0, food_demand[i]
            )

    return solution


# Genetic Algorithm
population_size = 50
generations = 100

population = [
    create_solution()
    for _ in range(population_size)
]

for generation in range(generations):

    # Select best solutions
    parents = selection(population)

    new_population = parents.copy()

    # Create new solutions
    while len(new_population) < population_size:

        parent1 = random.choice(parents)
        parent2 = random.choice(parents)

        child = crossover(parent1, parent2)

        child = mutation(child)

        new_population.append(child)

    population = new_population


# Best solution
best_solution = max(
    population,
    key=fitness
)

print("Best Resource Allocation:\n")

for i in range(members):

    print(
        "Member", i + 1,
        ": Water =", round(best_solution[i][0], 2),
        "Electricity =", round(best_solution[i][1], 2),
        "Food =", round(best_solution[i][2], 2)
    )

print("\nFitness =", round(fitness(best_solution), 4))
```
